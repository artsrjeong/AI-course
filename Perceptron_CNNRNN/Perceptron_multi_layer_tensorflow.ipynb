## 다층 Perceptron XOR Gate 학습(Activation Function Sigmoid 포함)

**tensorflow를 이용해서 학습**

![2 input, 1 output 학습](XOR_Gate.png)

**1. Tensorflow import, Training Set X,Y 설정**

```python
import tensorflow as tf
import numpy as np
# 1. 데이터 준비 (XOR 문제)
# 입력 데이터: [x1, x2]
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

# 정답(Target) 데이터 (XOR 결과)
y = np.array([[0],
              [1],
              [1],
              [0]])
```

**2. 모델 구성 : Dense Layer 2개)**

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(2, activation='sigmoid'),  # Hidden Layer (h1, h2)
    tf.keras.layers.Dense(1, activation='sigmoid')                     # Output Layer (y)
])
```

**3. 모델 컴파일**

```python
model.compile(optimizer='adam',   # adam은 경사하강법의 한 종류 
              loss='binary_crossentropy',  # 이진 분류에서는 MSE보다 binary_crossentropy가 효율적
              metrics=['accuracy'])
```

**4. 학습**

```python
model.fit(X, y, epochs=500, verbose=1)
```

**5. 예측**

```python
pred = model.predict(X)
print("XOR 예측 결과:")
print(np.round(pred))
```
