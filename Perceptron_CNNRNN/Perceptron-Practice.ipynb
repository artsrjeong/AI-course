## 단층 Perceptron AND Gate 학습(Activation Function 없이 간략화)

**Wt+1=Wt −𝛼∇𝐿(𝑊𝑡) 공식을 이용해서 weight와 bias 학습**

![2 input, 1 output 학습](And_Gate.png)

**1. Training Set X,Y 설정**

```python
    X = [
        [0, 0],
        [0, 1],
        [1, 0],
        [1, 1]
    ]
    Y = [0, 0, 0, 1]
```

**2. Hypter Parameter 설정(learning rate, epochs 등)**

```python
import random
import math
weights = [random.uniform(-1, 1) for _ in range(2)]
bias = random.uniform(-1, 1)
learning_rate = 0.01
epochs=5000
```

**3. 출력 예측 함수 작성**

```python
def predict(inputs):
    """출력 예측 (순전파)"""
    x1, x2 = inputs
    return x1 * weights[0] + x2 * weights[1] + bias    
```

**4. (출력-타겟) 오차를 계산하고 weight_new=weight_old - 학습률*error*입력 공식으로 weight 업데이트**

```python
for epoch in range(epochs):
    total_error = 0
    for inputs, label in zip(X, Y):

        # 1) 출력값 예측
        y_pred = predict(inputs)

        # 2) Error 계산 (Target - 예측값)
        error = y_pred-label

        # 3) Weight 업데이트
        weights[0] -= learning_rate * error * inputs[0]
        weights[1] -= learning_rate * error * inputs[1]

        # 4) Bias 업데이트
        bias -= learning_rate * error

        # 5) Loss 계산 (error^2)
        total_error += error ** 2

    # 1000회마다 오차 출력
    if epoch % 1000 == 0:
        print(f"Epoch {epoch}: Loss = {total_error / len(X):.4f}")
        print(weights, bias)
```

**5. 학습 결과 검증**

```python
print("\n--- 학습 결과 테스트 ---")
for inputs in X:
    result = predict(inputs)
    print(f"입력: {inputs}, 예측값: {result:.4f} -> 결과: {1 if result >= 0.5 else 0}")
```
